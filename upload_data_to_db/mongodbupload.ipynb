```python
import os
import certifi
from dotenv import load_dotenv
from pymongo import MongoClient
import pandas as pd
```

```python
load_dotenv()
```

```python
uri = os.getenv("MONGODB_URI")
db_name = os.getenv("DATABASE_NAME")
print("MongoDB URI:", uri)
print("Database Name:", db_name)
```

```python
ca = certifi.where()
```

```python
print("URI value:", uri)
print("Type:", type(uri))
print("DB Name value:", db_name)
print("Type:", type(db_name))
```

```python
# Validate URI scheme before connecting
if not uri or not (uri.startswith("mongodb://") or uri.startswith("mongodb+srv://")):
    raise ValueError(f"Invalid MongoDB URI: {uri}. It must start with 'mongodb://' or 'mongodb+srv://'")

try:
    client = MongoClient(uri, tlsCAFile=ca)
    db = client[db_name]
    print("Connected successfully")
except Exception as e:
    raise Exception(f"MongoDB Connection failed ; {e}")
```

```python
def upload_files_to_mongodb(datasets_dir_name):
    
    for file in os.listdir(datasets_dir_name):
        if file.endswith('.csv'):
            collection_name = file.split('.')[0]
            collection=db[collection_name]
            
            file_path = os.path.join(datasets_dir_name, file)
            print(f"processing file : {file_path}")
            
            df=pd.read_csv(file_path)
            
            df.columns = df.columns.str.strip()
            
            df=df.astype(str)
            
            data=df.to_dict(orient='records')
            
            if data:
                collection.insert_many(data)
                print(f"Inserted {len(data)} records into collection '{collection_name}'")
            else:
                print(f"No data to insert for file: {file_path}")


upload_files_to_mongodb(
    datasets_dir_name=r"E:\MLProject\PhisingClassifier\upload_data_to_db" )
```
